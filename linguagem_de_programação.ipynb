```python
filmes = ["filme 1", "filme 2", "filme 3", "filme 4", "filme 5"]
print("Bem vindo a classificacao do filme")
print("Você tem cinco filmes para classificar")
print(" Digite 0 a qualquer momento para parar.\n")

for filme in filmes:
  parar_programa = False
  while True:
    classificacao = input(f"como voce classifica o filme {filme} de 1 a 5 (ou 0 para parar): ")

    if classificacao == "0":
      print("Que pena")
      parar_programa = True
      break

    classificacao = int(classificacao)
    if classificacao < 1 or classificacao > 5:
      print("nota invalida")
    else:
      print(f"Voce classificou {filme} com {classificacao} estrelas.\n")
      print("obrigado")
      break

  if parar_programa:
    break
```

```python
#funções definidas pelo usuario
def calcular_desconto(valor, percentual):
  """calcula o valor final aplicando um desconto percentual."""
  if percentual < 0 or percentual > 100:
    return None #marcacao invalida (corrigido 'none' para 'None')
  desconto = valor * (percentual/100)
  return valor - desconto

def registrar_venda(produto, valor_final):
  """Imprime o registro de uma venda."""
  print(f">> venda registrada: {produto} por R$ {valor_final:.2f}")

#funcao lambda para arredondar
arredondar = lambda v: round(v, 2)
```

```python
#variáveis principais do PDV
total_vendas = 0
qtd_vendas = 0
```

```python
# Inicialização do caixa
total_vendas = 0
qtd_vendas = 0

# Loop principal do caixa
while True:
    print("\n----- MINHA LOJA - PDV -----")
    print("1 - Registrar venda")
    print("2 - Ver relatório do dia")
    print("3 - Encerrar caixa")

    opcao = input("Escolha uma opção: ")

    if opcao == "3":
        print("Caixa encerrado. Até amanhã!")
        break

    elif opcao == "1":
        produto = input("Nome do Produto: ")
        valor = float(input("Valor do produto: R$ "))
        percentual = float(input("Desconto (%): "))

        valor_final = calcular_desconto(valor, percentual)

        if valor_final is None:
            print("Desconto inválido! Use um valor entre 0 e 100.")

        else:
            valor_final = arredondar(valor_final)

            registrar_venda(produto, valor_final)

            total_vendas = total_vendas + valor_final
            qtd_vendas = qtd_vendas + 1

            print(f"Venda registrada: {produto}")
            print(f"Valor final: R$ {valor_final:.2f}")

    elif opcao == "2":
        print(f"\nVendas hoje: {qtd_vendas}")
        print(f"Total faturado: R$ {total_vendas:.2f}")

    else:
        print("Opção inválida. Tente novamente.")
```
